```python
np.random.seed(613) # gives us same random numbers, eve if cell/command is run multiple times
x = np.arange(50) # generates numbers in steps of one from 0 to 49
y = np.random.randint(0, 100,50) # generates 50 random integers between 0 and 100
```

```python
fig, ax = plt.subplots(figsize=(5, 3)) # determines figure size
ax.scatter(x,y) # general scatter plot (scattered points of x vs y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y) # instead of scatter plots the y-values are plotted agaainst the respective (same index) x values as bars
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y) # line plot of y vs x values
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y) # creating a histogram of the y values 
ax.set_title('Total growth over time') # title, fontsize etc can be adjusted
# ax. refers to the subplot one wants to adjust the characteristics of
ax.set_ylabel('Total growth') 
ax.set_xlabel('Years since start')
fig.tight_layout() # adjusts subplot parameters so that axes, labels, and titles fit neatly within the figure area without overlapping
```

```python
# fonts of different titles and labels are being defined

font1 = {'family':'sans-serif', 'color':'blue', 'size':20} 
font2 = {'family':'monospace','color':'green', 'size':14}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1) # by defining fontdict the color, family (=font) and size are automatically defined
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', color = 'indigo', linestyle = '--', linewidth = 2) 
ax.grid(axis = 'y', color = "blue", linewidth = 2, linestyle = '-.') # add grid in the background: style can be adjusted as well
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
# plotting two cuves in one pot (ax)
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right', 
          frameon = True, #add frame around the legend
          fontsize = 12, #change font size
          ncol = 2, #specify number of columns
          shadow = True )
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left',bbox_to_anchor =(1, 1) # set location of leged, anchor (1, 1) places a legend exactly at the top-right corner of the axes. 
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!", 
        ha='center', color = 'red', size = 20) 

fig.show()
```

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData ) # first to numbers define the position in the plot where the text is added
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes )
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure )

```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate("This is important!", xy=(10, 95), xytext=(20, 94), arrowprops = dict(arrowstyle = "wedge" , color = 'hotpink')) # drawung attention annotating) to a detail using an arrow 
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right', fontsize=12) # rotating the x-ticks by 45 deg
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig,(ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3)) # ax1 and ax2 instead of only one plot 9ax) per figure: 2 subplots
ax1.scatter(x1,y1) # plot and define properties of subplot 1
ax2.bar(x2,y2)  # plot and define properties of subplot 2
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
        ['ax2','ax3']],figsize=(7, 4), layout = "constrained") # constrained layout ensure no overlaps between subplots
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

```python
# creating two different style plots in one graph
x = np.array(["Luffy","Zoro", "Nami","Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))
# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, #our x values
            y2, #our y values
            yerr = y2_sd,
            fmt = "none", ecolor= "indigo",
            elinewidth= 4,
            capsize = 6,
            capthick= 4, errorevery= 2 ) #  errorevery= 2 plot every second errorbar
```

```python
from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
```

```python
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"
headers = {"User-Agent": "Mozilla/5.0","Accept": "image/*,*/*;q=0.8","Referer": "https://en.wikipedia.org/"}
response = requests.get(url, headers=headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35]) # adds image at certain position and size
ax_image.imshow(image)
ax_image.axis("off")
```

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
fig = plt.subplots(figsize=(10, 3)) # figure size
sns.lineplot(data=tips, # choose our dataset
             x='total_bill',
             y='tip')
sns.set_style('whitegrid') # set sns graph layout/background
tipgraph = sns.lineplot(data=tips,
                        x='total_bill',
                        y='tip',
                       color = 'hotpink', 
                        linestyle = '--',
                        linewidth = 3,
                        marker = 'o',
                        markerfacecolor = 'indigo')

tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')
```

```python
tipgraph = sns.scatterplot(data=tips, x='total_bill',
                           y='tip',
                           style = 'time', #set marker style as the dinner time 
                           hue = 'day', # color the markers by the weekday
                           palette = ['purple', 'hotpink', 'deepskyblue','yellowgreen']) # set possible colors for hue

tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)', ylabel='Tip Amount ($)')
```

```python
sns.pairplot(data = tips,hue = 'day') # pairplot gives you all combinations of the data rows as x vs y plots: pairwise relationships
```

```python
daysplot = sns.relplot(data=tips, x='total_bill',y="tip", 
                       hue="sex", col="day", kind="scatter",
                       col_wrap=2) # relplot plots the relationship between variables total bill and tip separated by the day as a scatterplot
```

```python
import plotly.graph_objects as go 
x1 = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y1 = np.array([110, 180, 240, 99, 220])
graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1)) # adds bar plot to existing graph figure that was just created in the line before

graph.update_layout(title="Pirate Scores", 
                    xaxis_title="Pirates",
                    yaxis_title="Score")
graph.show()
```

```python
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers', 
                           marker=dict(size=15,
                                       color='hotpink',
                                       opacity=1,
                                       line=dict(width=5, color='purple')))) # instead of bar plot the score is depicted by markers (whos style we define woht color, opacity, line etc)

graph.update_layout(title='Interactive Pirate Plot', xaxis_title='Pirates',
                    yaxis_title='Scores', width=500, height=500)

```

```python
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", on_bad_lines='skip')
df
```

```python
text = " ".join(each for each in df.quote)
wordcloud = WordCloud(background_color = 'white', colormap = 'inferno').generate(text) #generated wordcloud based on joined occurance (line above)
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, interpolation='bilinear') 
ax.axis("off")
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(["apple","banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

diagram = venn2_unweighted([A, B], set_labels = ('Fruits', 'Pies'),
                           set_colors = ('blue', 'red'),
                           alpha = 0.5) # create a venn diagramm of possible fruits and pies made of them
diagram.get_label_by_id("10").set_text("\n".join(A - B)) # label area that is only fruits
diagram.get_label_by_id("11").set_text("\n".join(A & B)) # label area that contains fruits that are also use din pies
diagram.get_label_by_id("01").set_text("\n".join(B - A)) # label area that is only pies which do not contain fruits 

plt.show()
```
